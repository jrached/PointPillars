```python
import torch 
from pointpillars.model.pointpillars import *
```

```python
voxel_size = [0.16, 0.16, 4]
point_cloud_range=[0, -39.68, -3, 69.12, 39.68, 1]
max_num_points=32
max_voxels=(16000, 40000)
```

```python
69.12 / 0.16
```

```python
39.68*2 / 0.16
```

```python
pillar_layer = PillarLayer(voxel_size=voxel_size, 
                                point_cloud_range=point_cloud_range, 
                                max_num_points=max_num_points, 
                                max_voxels=max_voxels)
pillar_encoder = PillarEncoder(voxel_size=voxel_size, 
                                    point_cloud_range=point_cloud_range, 
                                    in_channel=9, 
                                    out_channel=64)
backbone = Backbone(in_channel=64, 
                    out_channels=[64, 128, 256], 
                    layer_nums=[3, 5, 5])
neck = Neck(in_channels=[64, 128, 256], 
            upsample_strides=[1, 2, 4], 
            out_channels=[128, 128, 128])
```

```python
batch_points = [torch.rand(32, 4) for _ in range(4)]
```

```python
pillars, coors_batch, npoints_per_pillar = pillar_layer(batch_points)
```

```python
pillar_features = pillar_encoder(pillars, coors_batch, npoints_per_pillar)
```

```python
xs = backbone(pillar_features)
```
